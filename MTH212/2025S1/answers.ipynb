Q1(a)

Population variance is unknown, so use t-distribution to find confidence interval

Since sample size > 30, we can estimate the t-score using normal distribution.

```python
from sympy import *
from sympy.stats import *

alpha = 0.05
n = 49
x_bar = 2.45
s = 0.25

t = quantile(Normal("z", 0, 1))(1 - alpha/2)
print(f"t = {t.evalf(3)}")

upper = x_bar - t*s/sqrt(n)
lower = x_bar + t*s/sqrt(n)
print(f"95% confidence interval = [{upper.evalf(3)}, {lower.evalf(3)}]")
```

Q1(b)

Population variance is unknown, so use t-distribution to find confidence interval.

Sample size is 50.

Since sample size > 30, we can estimate the t-score using normal distribution.

Sample mean is $15000.

Upper bound of 99% confidence interval is $15364.87. 

Use upper bound and sample mean to find sample std dev.

```python
from sympy import *
from sympy.stats import *

n = 50
x_bar = 15000
u99 = 15364.87

t = lambda alpha: quantile(Normal("z", 0,1))(1-alpha/2)
print(f"t99 = {t(0.01).evalf(3)}")
print(f"t95 = {t(0.05).evalf(3)}")

s = sqrt(n)*(u99 - x_bar)/t(0.01)
print(f"s = {s.evalf(6)}")

upper = lambda alpha: x_bar + t(alpha)*s/sqrt(n)
lower = lambda alpha: x_bar - t(alpha)*s/sqrt(n)

print(f"99% confidence interval: [{lower(0.01).evalf(7)}, {upper(0.01).evalf(7)}]")
print(f"95% confidence interval: [{lower(0.05).evalf(7)}, {upper(0.05).evalf(7)}]")
```
