```python
import numpy as np 
from sympy import *
from sympy.stats import StudentT, Normal, density, quantile, cdf
```

Q1(a)

Population variance is unknown, so use t-distribution to find confidence interval. However, since sample size n > 30, we can approximate the t-distribution with a normal distribution.

```python
alpha = 0.05
n = 49
x_bar = 2.45
s = 0.25
t = quantile(Normal("z", 0, 1))(1 - alpha/2)
upper = x_bar - t*s/sqrt(n)
lower = x_bar + t*s/sqrt(n)
print(f"95% confidence interval = ({upper.evalf(3)}, {lower.evalf(3)})")
```

Q1(b)

Sample mean is $15000.

Uper bound of 99% confidence interval is $15364.87. 

Sample size is 50.

We can estimate t-score using z-score since sample size > 30.

```python
n = 50
x_bar = 15000
u99 = 15364.87

t = lambda alpha: quantile(Normal("z", 0,1))(1-alpha/2)
s = sqrt(n)*(u99 - x_bar)/t(0.01)
upper = lambda alpha: x_bar + t(alpha)*s/sqrt(n)
lower = lambda alpha: x_bar - t(alpha)*s/sqrt(n)

print(f"99% confidence interval: ({lower(0.01).evalf(7)}, {upper(0.01).evalf(7)})")
print(f"95% confidence interval: ({lower(0.05).evalf(7)}, {upper(0.05).evalf(7)})")
```
